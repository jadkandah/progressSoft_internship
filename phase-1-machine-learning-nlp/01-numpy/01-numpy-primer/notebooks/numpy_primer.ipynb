```python
python_features = features.tolist()
```

```python
python_features[0]
```

```python
np.min(features)
```

```python
min = 1000
for feature in python_features:
    for value in feature:
        if value < min:
            min = value
print("Minimum value in the dataset:", min)
```

```python
max = -1000
for feature in python_features:
    for value in feature:
        if value > max:
            max = value
print("Maximum value in the dataset:", max)
```

```python
total = 0
count = 0

for feature in python_features:
    for value in feature:
        total += value
        count += 1

average = total / count
print("Average value in the dataset:", average)
```

```python
# std = variance**0.5 -> variance = sum((value - average) ** 2 / count

sum_squared_diff = 0

for feature in python_features:
    for value in feature:
        sum_squared_diff += (value - average) ** 2

variance = sum_squared_diff / count
std = variance ** 0.5

print("Standard Deviation of the dataset:", std)
```

```python
# finding the mean of each feature using python

num_rows = len(python_features)
num_cols = len(python_features[0])

feature_means = []

for col in range(num_cols):
    col_sum = 0
    for row in range(num_rows):
        col_sum += python_features[row][col]
    feature_means.append(col_sum / num_rows)

print("Feature Means:", feature_means)
```

```python
# same for col max, min, std
```

```python
# col max

col_maxes = []

for col in range(num_cols):
    col_max = python_features[0][col]
    for row in range(1, num_rows):
        if python_features[row][col] > col_max:
            col_max = python_features[row][col]
    col_maxes.append(col_max)

print("Column Maxes:", col_maxes)
```

```python
# Numerical operations using python

# features + 1

added_features = []
for feature in python_features:
    added_row = []
    for value in feature:
        added_row.append(value + 1)
    added_features.append(added_row)

print("Added Features:", added_features)
```

### Compare the execution time systematically (through multiple repeats and then taking the mean and std) between writing the previous computations in pure Python and in NumPy. In a text/markdown block (if you have a notebook) or in a comment (if you have a script), write your observations and why is NumPy faster than pure Python?

```python
import timeit


def numpy_statistics():
    return (
        np.min(features),
        np.max(features),
        np.mean(features),
        np.std(features),
    )


def python_statistics():
    total = 0
    count = 0
    minimum = python_features[0][0]
    maximum = python_features[0][0]

    for row in python_features:
        for value in row:
            total += value
            count += 1

            if value < minimum:
                minimum = value

            if value > maximum:
                maximum = value

    mean = total / count
    squared_difference_sum = 0

    for row in python_features:
        for value in row:
            squared_difference_sum += (value - mean) ** 2

    standard_deviation = (squared_difference_sum / count) ** 0.5

    return minimum, maximum, mean, standard_deviation


repeats = 10
executions_per_repeat = 10000

numpy_times = timeit.repeat(
    numpy_statistics,
    repeat=repeats,
    number=executions_per_repeat,
)

python_times = timeit.repeat(
    python_statistics,
    repeat=repeats,
    number=executions_per_repeat,
)

numpy_times = np.array(numpy_times) / executions_per_repeat
python_times = np.array(python_times) / executions_per_repeat

print("NumPy result:", numpy_statistics())
print("Python result:", python_statistics())

print(
    f"NumPy:  {numpy_times.mean():.8f} ± "
    f"{numpy_times.std():.8f} seconds"
)

print(
    f"Python: {python_times.mean():.8f} ± "
    f"{python_times.std():.8f} seconds"
)

print(
    f"Speedup: {python_times.mean() / numpy_times.mean():.2f}x"
)
```

### From the results it shows that numpy doing the same processes took a fourth of the time it takes for the same process to happen when impemented in pure python doing the same processes and I assume the reason for that is the way is because numpy stores only the same data type for all items in it while using pure python which stores each item as an entire object and has a pointer to it in a list so this will take time at each item to go to the pointed to object and doing an operation on it and then storing a new pointer again.
