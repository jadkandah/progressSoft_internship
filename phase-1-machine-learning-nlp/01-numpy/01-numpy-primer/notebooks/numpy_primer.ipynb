### Compare the execution time systematically (through multiple repeats and then taking the mean and std) between writing the previous computations in pure Python and in NumPy. In a text/markdown block (if you have a notebook) or in a comment (if you have a script), write your observations and why is NumPy faster than pure Python?

```python
import timeit


def numpy_statistics():
    return (
        np.min(features),
        np.max(features),
        np.mean(features),
        np.std(features),
    )


def python_statistics():
    total = 0
    count = 0
    minimum = python_features[0][0]
    maximum = python_features[0][0]

    for row in python_features:
        for value in row:
            total += value
            count += 1

            if value < minimum:
                minimum = value

            if value > maximum:
                maximum = value

    mean = total / count
    squared_difference_sum = 0

    for row in python_features:
        for value in row:
            squared_difference_sum += (value - mean) ** 2

    standard_deviation = (squared_difference_sum / count) ** 0.5

    return minimum, maximum, mean, standard_deviation


repeats = 10
executions_per_repeat = 10000

numpy_times = timeit.repeat(
    numpy_statistics,
    repeat=repeats,
    number=executions_per_repeat,
)

python_times = timeit.repeat(
    python_statistics,
    repeat=repeats,
    number=executions_per_repeat,
)

numpy_times = np.array(numpy_times) / executions_per_repeat
python_times = np.array(python_times) / executions_per_repeat

print("NumPy result:", numpy_statistics())
print("Python result:", python_statistics())

print(
    f"NumPy:  {numpy_times.mean():.8f} ± "
    f"{numpy_times.std():.8f} seconds"
)

print(
    f"Python: {python_times.mean():.8f} ± "
    f"{python_times.std():.8f} seconds"
)

print(
    f"Speedup: {python_times.mean() / numpy_times.mean():.2f}x"
)
```

### From the results it shows that numpy doing the same processes took a fourth of the time it took for the same process to happen when impemented in pure python doing the same processes and I assume the reason for that is the way is because numpy stores only the same data type for all items in it while using pure python which stores each item as an entire object and has a pointer to it in a list so this will take time at each item to go to the pointed to object and doing an operation on it and then storing a new pointer again.

```python
from itertools import product

import matplotlib.pyplot as plt
import timeit


def numpy_statistics_for(values):
    """Calculate the four statistics with NumPy."""
    return (
        np.min(values),
        np.max(values),
        np.mean(values),
        np.std(values),
    )


def python_statistics_for(values):
    """Calculate the same four statistics with Python loops."""
    if not values:
        raise ValueError("values cannot be empty")

    total = 0.0
    minimum = values[0]
    maximum = values[0]

    for value in values:
        total += value
        if value < minimum:
            minimum = value
        if value > maximum:
            maximum = value

    mean = total / len(values)
    squared_difference_sum = 0.0
    for value in values:
        squared_difference_sum += (value - mean) ** 2

    standard_deviation = (squared_difference_sum / len(values)) ** 0.5
    return minimum, maximum, mean, standard_deviation


# Use the same values and data type for both implementations. Input creation is
# deliberately outside the timed region. Larger inputs repeat the Iris values.
base_values = np.asarray(features, dtype=np.float64).ravel()
input_sizes = [10, 100, 1_000, 10_000]
execution_counts = [1, 10, 100, 1_000]
repeats = 7
benchmark_rows = []

# Test every data size with every executions-per-repeat setting.
for size, executions in product(input_sizes, execution_counts):
    numpy_input = np.resize(base_values, size)
    python_input = numpy_input.tolist()

    # Verify correctness before comparing speed.
    numpy_result = numpy_statistics_for(numpy_input)
    python_result = python_statistics_for(python_input)
    assert np.allclose(numpy_result, python_result, rtol=1e-12, atol=1e-12)

    numpy_samples = []
    python_samples = []

    # Warm up both functions once, then alternate timing order between repeats
    # to reduce systematic first/second-run bias.
    numpy_statistics_for(numpy_input)
    python_statistics_for(python_input)
    for repeat_index in range(repeats):
        if repeat_index % 2 == 0:
            numpy_elapsed = timeit.timeit(
                lambda: numpy_statistics_for(numpy_input), number=executions
            )
            python_elapsed = timeit.timeit(
                lambda: python_statistics_for(python_input), number=executions
            )
        else:
            python_elapsed = timeit.timeit(
                lambda: python_statistics_for(python_input), number=executions
            )
            numpy_elapsed = timeit.timeit(
                lambda: numpy_statistics_for(numpy_input), number=executions
            )

        # Keep total batch time to show the cost of increasing executions.
        numpy_samples.append(numpy_elapsed)
        python_samples.append(python_elapsed)

    numpy_samples = np.asarray(numpy_samples)
    python_samples = np.asarray(python_samples)
    paired_speedups = python_samples / numpy_samples

    benchmark_rows.append({
        "Values": size,
        "Executions per repeat": executions,
        "NumPy total mean (ms)": numpy_samples.mean() * 1_000,
        "NumPy total std (ms)": numpy_samples.std(ddof=1) * 1_000,
        "Python total mean (ms)": python_samples.mean() * 1_000,
        "Python total std (ms)": python_samples.std(ddof=1) * 1_000,
        "Speedup mean": paired_speedups.mean(),
        "Speedup std": paired_speedups.std(ddof=1),
    })

benchmark_results = pd.DataFrame(benchmark_rows)
print(benchmark_results.to_string(index=False, float_format=lambda value: f"{value:.4f}"))

# One chart per size makes the small-input crossover easy to see.
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.ravel()
bar_width = 0.36
positions = np.arange(len(execution_counts))

for axis, size in zip(axes, input_sizes):
    subset = benchmark_results[
        benchmark_results["Values"] == size
    ].sort_values("Executions per repeat")

    axis.bar(
        positions - bar_width / 2, subset["NumPy total mean (ms)"],
        width=bar_width, yerr=subset["NumPy total std (ms)"],
        capsize=3, label="NumPy", color="tab:blue",
    )
    axis.bar(
        positions + bar_width / 2, subset["Python total mean (ms)"],
        width=bar_width, yerr=subset["Python total std (ms)"],
        capsize=3, label="Pure Python", color="tab:orange",
    )

    axis.set_title(f"{size:,} values", fontweight="bold")
    axis.set_xticks(positions, [f"{count:,}" for count in execution_counts])
    axis.set_xlabel("Executions per repeat")
    axis.set_ylabel("Mean total time per repeat (ms)")
    axis.set_yscale("log")
    axis.grid(axis="y", which="both", alpha=0.25)
    axis.legend()

fig.suptitle(
    "NumPy vs pure Python: small inputs and increasing executions", fontsize=15
)
fig.tight_layout()
plt.show()
```

Python is faster in stuff where there are different data types in same array if we do this using NumPy and specify dtype as object it is even slower than Python Objects due to overhead of transitioning from NumPy to Python alot.

### Pure Python (Element-wise Addition)
- Start executing the Python bytecode.
- Enter the interpreter loop.
- Fetch the next iteration of the loop.
- Read the pointer to the first operand from the first list.
- Dereference the pointer to obtain the first PyObject.
- Read the pointer to the second operand from the second list.
- Dereference the pointer to obtain the second PyObject.
- Check the type (ob_type) of both objects.
- Dispatch the + operator through PyNumber_Add().
- Resolve the correct addition function.
- Perform the arithmetic operation.
- Allocate a new Python object for the result.
- Initialize the new object's metadata (reference count, type, value).
- Store a pointer to the new object in the output list.
- Repeat steps 3–14 for every remaining element.
- Return the completed output list.

### NumPy (Element-wise Addition)
- Start executing the Python bytecode.
- Call ndarray.__add__().
- Enter the NumPy C API.
- Verify that both operands are valid ndarray objects.
- Check the arrays' data types (dtype).
- Check the arrays' shapes.
- Verify broadcasting compatibility.
- Resolve the appropriate universal function (numpy.add).
- Select the optimized inner loop corresponding to the arrays' dtype.
- Allocate contiguous memory for the output array.
- Initialize the output array's metadata (shape, strides, dtype, flags, etc.).
- Transfer execution from the Python interpreter to the compiled C loop.
- Iterate through the contiguous memory, performing the addition directly on raw values.
- Store each result directly into the output array's memory.
- Exit the compiled C loop.
- Return the completed ndarray to Python.
